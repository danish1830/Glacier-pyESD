```python
import pandas as pd
from docx import Document

# Read and prepare data
df = pd.read_csv(r"C:\Users\danis\Desktop\Danish\Thesis Model Results\model_metrics_bs.csv")
df = df.round(3)

# Create a new Word doc
doc = Document()
doc.add_heading('Model Results', level=1)

# Add table
table = doc.add_table(rows=1, cols=len(df.columns))
table.style = 'Table Grid'

# Add header row
hdr_cells = table.rows[0].cells
for i, col_name in enumerate(df.columns):
    hdr_cells[i].text = col_name

# Add data rows
for _, row in df.iterrows():
    row_cells = table.add_row().cells
    for i, value in enumerate(row):
        row_cells[i].text = str(value)

# Save the Word document
doc.save(r"C:\Users\danis\Desktop\Danish\Thesis Model Results\model_results_bs.docx")
```
