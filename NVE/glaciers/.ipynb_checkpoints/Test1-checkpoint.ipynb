```python
import pandas as pd  # Data manipulation
import numpy as np   # Numerical computing
import matplotlib.pyplot as plt  # Plotting
import seaborn as sns  # Statistical visualization
import glob
import os
from sklearn.linear_model import LinearRegression  # For regression models
```

```python
folder_path = "C:/Users/danis/Desktop/NVE/NVE/glaciers/Mass Balance/"
file_list = glob.glob(os.path.join(folder_path, "*.csv"))  # Find all CSV files
```

```python
def read_mass_balance(file):
    # Read the CSV file
    data = pd.read_csv(file, delimiter=';', dtype=str)  # Read as string like R
    data.columns = data.columns.str.strip()  # Trim spaces in column names

  # Convert to numeric (handling commas in decimals)
    for col in ["Year", "Ba", "Bw", "Bs"]:
        data[col] = data[col].str.replace(",", ".").astype(float)

    # Add Glacier name from filename
    data["Glacier"] = os.path.basename(file).split("_")[-1].split(".")[0]

    return data

# Read all CSV files and combine into one DataFrame
glacier_data = pd.concat([read_mass_balance(f) for f in file_list if read_mass_balance(f) is not None], ignore_index=True)
```

```python
glacier_data = glacier_data.sort_values(by=["Glacier", "Year"])

# Apply moving average (rolling mean)
glacier_data["Bw_smoothed"] = glacier_data.groupby("Glacier")["Bw"].transform(lambda x: x.rolling(window=5, center=True).mean())
glacier_data["Bs_smoothed"] = glacier_data.groupby("Glacier")["Bs"].transform(lambda x: x.rolling(window=5, center=True).mean())
```

```python
plt.figure(figsize=(10, 5))
sns.lineplot(data=glacier_data, x="Year", y="Bw_smoothed", hue="Glacier", linewidth=1.2)

plt.title("Winter Mass Balance Trends: Rolling Mean")
plt.xlabel("Year")
plt.ylabel("Bw (m w.e.)")
plt.legend(title="Glacier")
plt.grid(True)
plt.show()
```
